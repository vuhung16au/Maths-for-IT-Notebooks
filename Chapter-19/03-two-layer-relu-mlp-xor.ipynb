## Mathematical Background

The XOR dataset consists of four points:
- $(0, 0) \to 0$ and $(1, 1) \to 0$
- $(0, 1) \to 1$ and $(1, 0) \to 1$

No single straight line can separate class 0 from class 1. A 2-layer MLP with hidden dimension $H$ computes:
$$Z_1 = X W_1 + \mathbf{b}_1, \qquad H_{\text{act}} = \operatorname{ReLU}(Z_1) = \max(0, Z_1)$$
$$Z_2 = H_{\text{act}} W_2 + \mathbf{b}_2, \qquad \widehat{\mathbf{p}} = \sigma(Z_2)$$

Using Binary Cross-Entropy loss $L$, the backpropagation gradients are:
$$dZ_2 = \frac{1}{n}(\widehat{\mathbf{p}} - \mathbf{y}), \qquad dW_2 = H_{\text{act}}^T dZ_2, \qquad d\mathbf{b}_2 = \sum_{\text{rows}} dZ_2$$
$$dH_{\text{act}} = dZ_2 W_2^T, \qquad dZ_1 = dH_{\text{act}} \odot \mathbb{I}(Z_1 > 0)$$
$$dW_1 = X^T dZ_1, \qquad d\mathbf{b}_1 = \sum_{\text{rows}} dZ_1$$
