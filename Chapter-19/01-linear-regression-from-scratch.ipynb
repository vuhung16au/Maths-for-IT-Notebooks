## Mathematical Background

Given $n$ data pairs $(x_i, y_i)$, univariate linear regression predicts:
$$\widehat{y}_i = w x_i + b$$
and minimises the **Mean Squared Error (MSE)**:
$$L(w, b) = \frac{1}{n} \sum_{i=1}^{n} (\widehat{y}_i - y_i)^2$$

The partial derivatives of $L(w, b)$ are:
$$\frac{\partial L}{\partial w} = \frac{2}{n} \sum_{i=1}^{n} (\widehat{y}_i - y_i) x_i, \qquad \frac{\partial L}{\partial b} = \frac{2}{n} \sum_{i=1}^{n} (\widehat{y}_i - y_i)$$
