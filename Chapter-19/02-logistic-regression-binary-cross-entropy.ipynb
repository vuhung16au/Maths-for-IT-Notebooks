## Mathematical Background

Logistic regression maps a linear score $z = \mathbf{w} \cdot \mathbf{x} + b$ to a probability $\widehat{p} \in (0, 1)$ via the sigmoid function:
$$\widehat{p} = \sigma(z) = \frac{1}{1 + e^{-z}}$$

It minimises **Binary Cross-Entropy (BCE)** loss over $n$ samples:
$$L(\mathbf{w}, b) = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \ln(\widehat{p}_i) + (1 - y_i) \ln(1 - \widehat{p}_i) \right]$$

Because $\sigma'(z) = \sigma(z)(1 - \sigma(z))$, the gradient simplifies cleanly to:
$$\nabla_{\mathbf{w}} L = \frac{1}{n} X^T (\widehat{\mathbf{p}} - \mathbf{y}), \qquad \frac{\partial L}{\partial b} = \frac{1}{n} \sum_{i=1}^{n} (\widehat{p}_i - y_i)$$
