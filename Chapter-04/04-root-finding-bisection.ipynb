```python
# Worked example: x^2 - 5x + 6 = 0 has roots at 2 and 3
f1 = lambda x: x**2 - 5 * x + 6
r1 = bisection(f1, 1.5, 2.5)
r2 = bisection(f1, 2.5, 3.5)
assert math.isclose(r1, 2.0, abs_tol=1e-6)
assert math.isclose(r2, 3.0, abs_tol=1e-6)

# Approximate sqrt(2)
sqrt2 = bisection(lambda x: x**2 - 2.0, 1.0, 2.0, tol=1e-10)
assert math.isclose(sqrt2, math.sqrt(2.0), abs_tol=1e-9)
print(f'bisection sqrt(2) = {sqrt2:.10f}')
print('All Chapter 4.4 checks passed!')
```
