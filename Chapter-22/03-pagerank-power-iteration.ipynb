### Mathematical Context

Given a directed graph on $n$ nodes with adjacency matrix $M$ ($M_{ij} = 1$ if $i \to j$), the **column-stochastic transition matrix** $P$ has entries:
$$P_{ji} = \begin{cases} \dfrac{M_{ij}}{\text{out\_deg}(i)} & \text{if } \text{out\_deg}(i) > 0, \\[6pt] \dfrac{1}{n} & \text{if } \text{out\_deg}(i) = 0 \text{ (dangling node).} \end{cases}$$

Starting from $\mathbf{r}^{(0)} = \frac{1}{n}\mathbf{1}$, **PageRank power iteration** updates:
$$\mathbf{r}^{(k+1)} = d\, P\, \mathbf{r}^{(k)} + \frac{1 - d}{n}\mathbf{1}$$
until $\|\mathbf{r}^{(k+1)} - \mathbf{r}^{(k)}\|_1 < \text{tol}$.
