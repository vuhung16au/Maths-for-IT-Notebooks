### Mathematical Context & Worked Example

An $n$-vertex graph $G = (V, E)$ with vertices indexed $0, 1, \dots, n-1$ can be stored as an $n \times n$ **adjacency matrix** $M$:
$$M_{ij} = \begin{cases} w(i, j) & \text{if } (i, j) \in E, \\ 0 & \text{otherwise.} \end{cases}$$

For the textbook's router network ($A=0, B=1, C=2, D=3$) with undirected links $AB=2, AC=5, BC=1, BD=4, CD=1$:
$$M = \begin{bmatrix} 0 & 2 & 5 & 0 \\ 2 & 0 & 1 & 4 \\ 5 & 1 & 0 & 1 \\ 0 & 4 & 1 & 0 \end{bmatrix}.$$

A **tree** on $n$ vertices is a connected undirected graph with **no cycles**, which holds if and only if it is connected and has exactly $n - 1$ edges.
