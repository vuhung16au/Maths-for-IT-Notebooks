### Mathematical Context & Worked Example

Given a graph with **non-negative** edge weights $w(u, v) \ge 0$ and a source vertex $s$, Dijkstra's algorithm maintains tentative distances $d[v]$ (initialized to $d[s] = 0$ and $d[v] = \infty$ for $v \neq s$) and repeatedly settles the unvisited vertex $u$ with minimum $d[u]$, relaxing each outgoing edge $(u, v)$:
$$d[v] \leftarrow \min\bigl(d[v],\; d[u] + w(u, v)\bigr).$$

In the textbook's router network ($A=0, B=1, C=2, D=3$):
- Start at $A$: $d = [0, \infty, \infty, \infty]$. Settle $A \implies d[B]=2, d[C]=5$.
- Settle $B$ ($d[B]=2$): update $d[C] = \min(5, 2+1) = 3$, $d[D] = 2+4 = 6$.
- Settle $C$ ($d[C]=3$): update $d[D] = \min(6, 3+1) = 4$.
- Optimal route: $A \to B \to C \to D$ with total cost $4$.
