```python
# Toy 4D embeddings: [royalty, gender_fem, youth, fruit]
embeddings = {
    "man": np.array([0.1, 0.1, 0.2, 0.0]),
    "woman": np.array([0.1, 0.9, 0.2, 0.0]),
    "king": np.array([0.9, 0.1, 0.2, 0.0]),
    "queen": np.array([0.9, 0.9, 0.2, 0.0]),
    "prince": np.array([0.85, 0.1, 0.85, 0.0]),
    "princess": np.array([0.85, 0.9, 0.85, 0.0]),
    "apple": np.array([0.0, 0.0, 0.1, 0.95]),
}

# Test king - man + woman -> queen
target = analogy_vector(embeddings["man"], embeddings["woman"], embeddings["king"])
word, sim = nearest_word(target, embeddings, exclude={"man", "woman", "king"})
assert word == "queen", f"Expected 'queen', got '{word}'"
assert math.isclose(sim, 1.0, abs_tol=1e-9)

# Test prince - man + woman -> princess
target2 = analogy_vector(embeddings["man"], embeddings["woman"], embeddings["prince"])
word2, _ = nearest_word(target2, embeddings, exclude={"man", "woman", "prince"})
assert word2 == "princess"

# Unrelated word 'apple' should have low similarity to 'king'
assert cosine_similarity(embeddings["king"], embeddings["apple"]) < 0.1

print(f"Analogy 'king - man + woman' -> '{word}' (similarity={sim:.4f})")
print("All Chapter 7.3 checks passed!")
```
