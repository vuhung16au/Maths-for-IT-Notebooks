### Mathematical Context

For an event $A$ (e.g., fraudulent transaction) and evidence $B$ (e.g., detector flag), **Bayes' theorem** states:
$$P(A \mid B) = \frac{P(B \mid A)\,P(A)}{P(B)},$$
where by the **Law of Total Probability**:
$$P(B) = P(B \mid A)\,P(A) + P(B \mid A^c)\,P(A^c).$$

In the chapter's worked example:
- Base rate (prior): $P(\text{fraud}) = 0.10$
- Sensitivity (true positive rate): $P(\text{flag} \mid \text{fraud}) = 0.90$
- False positive rate: $P(\text{flag} \mid \text{legit}) = 0.05$

Among $1{,}000$ transactions, there are $90$ true flags and $45$ false flags, so $P(\text{fraud} \mid \text{flag}) = 90 / 135 = 2/3 \approx 0.667$.
