### Mathematical Context

$$\mathbb E[X] = \sum_i x_i p_i, \qquad \operatorname{Var}(X) = \sum_i (x_i - \mathbb E[X])^2 p_i.$$

In the chapter's worked example, a service receives $0, 1,$ or $2$ requests with probabilities $0.2, 0.5, 0.3$, yielding:
$$\mathbb E[X] = 0(0.2) + 1(0.5) + 2(0.3) = 1.1.$$
Notice that $\mathbb E[X]$ is the long-run average (balance point), even though $1.1$ is not a possible single outcome.
