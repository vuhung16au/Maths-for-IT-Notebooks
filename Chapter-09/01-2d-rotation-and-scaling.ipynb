```python
# Self-check tests matching Chapter 9 worked examples
R90 = rotation_matrix_2d(math.pi / 2.0)
assert np.allclose(R90, [[0.0, -1.0], [1.0, 0.0]], atol=1e-12)

v = np.array([3.0, 2.0])
rv = R90 @ v
assert np.allclose(rv, [-2.0, 3.0])
assert math.isclose(float(np.linalg.norm(rv)), math.sqrt(13.0))

# Composition order: scale then rotate vs rotate then scale
S = scaling_matrix_2d(2.0, 1.0)
u = np.array([1.0, 1.0])
assert np.allclose((R90 @ S) @ u, [-1.0, 2.0])
assert np.allclose((S @ R90) @ u, [-2.0, 1.0])

# Shear on unit square
unit_square = np.array([[0.0, 0.0], [1.0, 0.0], [1.0, 1.0], [0.0, 1.0]])
H = shear_matrix_2d(kx=0.7, ky=0.0)
sheared = transform_points_2d(H, unit_square)
assert np.allclose(sheared, [[0.0, 0.0], [1.0, 0.0], [1.7, 1.0], [0.7, 1.0]])

print("All Chapter 9.1 checks passed!")
```
