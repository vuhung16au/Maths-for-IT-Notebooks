```python
# Self-check tests
u = np.array([1.0, 2.0, 3.0])
assert math.isclose(cosine_similarity(u, 5.0 * u), 1.0, abs_tol=1e-9)  # scale invariance
assert math.isclose(cosine_similarity(np.array([1.0, 0.0]), np.array([0.0, 1.0])), 0.0, abs_tol=1e-9)
assert math.isclose(cosine_similarity(u, -u), -1.0, abs_tol=1e-9)

# Ranking deterministic batch
query = np.array([1.0, 0.8, 0.2, 0.0])
candidates = np.array([
    [0.0, 0.0, 0.1, 1.0],   # 0: unrelated
    [2.0, 1.6, 0.4, 0.0],   # 1: exact same direction (scaled)
    [-1.0, -0.8, -0.2, 0.0],# 2: opposite direction
    [0.9, 0.7, 0.3, 0.1],   # 3: very close
])
ranked = rank_candidates(query, candidates)
assert [idx for idx, _ in ranked] == [1, 3, 0, 2]
assert math.isclose(ranked[0][1], 1.0, abs_tol=1e-9)
assert math.isclose(ranked[-1][1], -1.0, abs_tol=1e-9)

print("Ranked candidates:", [(i, round(s, 4)) for i, s in ranked])
print("All Chapter 10.1 checks passed!")
```
