$$\operatorname{RSS}(\mathbf{w}) = \lVert A\mathbf{w} - \mathbf{y} \rVert_2^2.$$

```python
# Self-check 1: Exact line y = 2x + 1
x_exact = np.array([0.0, 1.0, 2.0, 3.0])
y_exact = 2.0 * x_exact + 1.0
m_ne, c_ne = fit_line_normal_equations(x_exact, y_exact)
m_ls, c_ls = fit_line_lstsq(x_exact, y_exact)
assert math.isclose(m_ne, 2.0) and math.isclose(c_ne, 1.0)
assert math.isclose(m_ls, 2.0) and math.isclose(c_ls, 1.0)
assert math.isclose(residual_sum_of_squares(x_exact, y_exact, m_ls, c_ls), 0.0, abs_tol=1e-12)

# Self-check 2: Noisy synthetic data with rng = np.random.default_rng(42)
rng = np.random.default_rng(42)
x = np.linspace(0.0, 10.0, 25)
y = 1.5 * x + 2.0 + rng.normal(0.0, 0.5, size=x.shape)

m1, c1 = fit_line_normal_equations(x, y)
m2, c2 = fit_line_lstsq(x, y)
assert math.isclose(m1, m2, rel_tol=1e-10)
assert math.isclose(c1, c2, rel_tol=1e-10)

# Verify orthogonality of residual r = y - A w_hat to col(A): A.T @ r == 0
A = np.column_stack([x, np.ones_like(x)])
r = y - (m2 * x + c2)
assert np.allclose(A.T @ r, np.zeros(2), atol=1e-10)

rss_opt = residual_sum_of_squares(x, y, m2, c2)
assert rss_opt < residual_sum_of_squares(x, y, m2 + 0.2, c2)

print(f"Fitted line: y = {m2:.4f}x + {c2:.4f}, RSS = {rss_opt:.4f}")
print("All Chapter 10.4 checks passed!")
```
