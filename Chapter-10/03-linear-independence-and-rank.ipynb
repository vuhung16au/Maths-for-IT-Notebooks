## Mathematical Context

The **rank** of a matrix is the number of linearly independent rows (equal to the number of linearly independent columns).
For a linear system $A\mathbf{x} = \mathbf{b}$ with $A \in \mathbb{R}^{m \times n}$:
- If $\operatorname{rank}(A) < \operatorname{rank}([A \mid \mathbf{b}])$, the system is **inconsistent** (0 solutions).
- If $\operatorname{rank}(A) = \operatorname{rank}([A \mid \mathbf{b}]) = n$, the system has a **unique** solution.
- If $\operatorname{rank}(A) = \operatorname{rank}([A \mid \mathbf{b}]) < n$, the system has **infinitely many** solutions.
