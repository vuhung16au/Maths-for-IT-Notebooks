### Mathematical Background

A definite integral $\int_a^b f(x)\,dx$ represents the signed area under $f(x)$ on $[a, b]$. Partitioning $[a, b]$ into $n$ subintervals of width $\Delta x = \frac{b - a}{n}$ gives the **Riemann sum**:

$$S_n = \sum_{i=0}^{n-1} f(x_i^*) \Delta x$$

where $x_i^*$ is chosen as:
- **Left endpoint:** $x_i^* = a + i\Delta x$
- **Right endpoint:** $x_i^* = a + (i + 1)\Delta x$
- **Midpoint:** $x_i^* = a + (i + 0.5)\Delta x$

In probability and statistics (Section 15.4), $P(a \le X \le b) = \int_a^b p(x)\,dx$ for a continuous density $p(x) \ge 0$.

```python
# 1. Exercise 15.5: Uniform density p(x) = 1/4 on [0, 4], P(1 <= X <= 3) = 0.5
uniform_pdf = lambda x: np.full_like(x, 0.25, dtype=np.float64)
p_uniform = riemann_sum(uniform_pdf, 1.0, 3.0, n=100, method='midpoint')
assert math.isclose(p_uniform, 0.5, rel_tol=1e-9)

# 2. Integral of exp(-x^2) on [0, 1] (true value ~ 0.746824132812427)
gauss_fn = lambda x: np.exp(-(x ** 2))
true_val = 0.746824132812427
left_100 = riemann_sum(gauss_fn, 0.0, 1.0, n=100, method='left')
right_100 = riemann_sum(gauss_fn, 0.0, 1.0, n=100, method='right')
mid_100 = riemann_sum(gauss_fn, 0.0, 1.0, n=100, method='midpoint')
assert abs(mid_100 - true_val) < abs(left_100 - true_val)
assert abs(mid_100 - true_val) < abs(right_100 - true_val)
assert math.isclose(mid_100, true_val, abs_tol=1e-5)

# 3. Standard normal 1-sigma probability P(-1 <= X <= 1) ~ 0.68268949
p_1sigma = estimate_gaussian_probability(-1.0, 1.0, n=1000)
assert math.isclose(p_1sigma, 0.68268949, abs_tol=1e-5)

print('All Chapter 15.3 assertions passed!')
```
