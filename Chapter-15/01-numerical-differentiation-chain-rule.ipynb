### Mathematical Background

The derivative $f'(x)$ measures the instantaneous rate of change (tangent slope) of $f$ at $x$. Numerically, the **central difference** formula provides an $O(h^2)$-accurate approximation:

$$f'(x) \approx \frac{f(x + h) - f(x - h)}{2h}$$

For composite functions $h(x) = f(g(x))$, the **chain rule** multiplies local sensitivities:

$$h'(x) = f'(g(x)) \cdot g'(x)$$

In Section 15.3.1 (Worked Example), for a sigmoid neuron $h(x) = \sigma(2x + 1)$ where $\sigma(u) = \frac{1}{1 + e^{-u}}$:

$$h'(x) = 2\sigma(2x + 1)\bigl(1 - \sigma(2x + 1)\bigr)$$
