### Mathematical Background

Given a data matrix $X \in \mathbb{R}^{n \times d}$, **PCA** (Section 17.3) finds orthogonal directions of maximum variance:

1. **Center:** $\tilde{X} = X - \mu$, where $\mu = \frac{1}{n}\sum_{i=1}^n \mathbf{x}_i$.
2. **SVD:** $\tilde{X} = U \Sigma V^T$ (`np.linalg.svd(X_centered, full_matrices=False)`).
3. **Principal components:** The top $k$ rows of $V^T$ form $W_k \in \mathbb{R}^{k \times d}$.
4. **Projection:** $Z = \tilde{X} W_k^T \in \mathbb{R}^{n \times k}$.
5. **Explained variance ratio:** $\frac{\sigma_i^2}{\sum_{j} \sigma_j^2}$.

```python
# 1. Figure 17.4-style correlated 2D point cloud
X_fig = np.array([
    [-1.2, -0.7], [-0.8, -0.4], [-0.3, -0.3],
    [0.2, 0.1], [0.6, 0.5], [1.1, 0.6], [1.4, 0.9]
], dtype=np.float64)
X_1d, comps_1d, evr_1d, _ = pca_svd(X_fig, n_components=1)
assert X_1d.shape == (7, 1)
assert comps_1d.shape == (1, 2)
assert np.isclose(np.linalg.norm(comps_1d[0]), 1.0)
assert evr_1d[0] > 0.98, f'Expected PC1 to capture >98% variance, got {evr_1d[0]}'

# 2. Full 2-component decomposition & orthogonality
X_2d, comps_2d, evr_2d, mean_2d = pca_svd(X_fig, n_components=2)
assert np.allclose(comps_2d @ comps_2d.T, np.eye(2), atol=1e-7)
assert np.isclose(np.sum(evr_2d), 1.0)
# Reconstruction from all components equals original X_fig
X_recon = X_2d @ comps_2d + mean_2d
assert np.allclose(X_recon, X_fig, atol=1e-7)

print(f'PC1 explained variance ratio: {evr_1d[0]:.4f}')
print('All Chapter 17.3 assertions passed!')
```
