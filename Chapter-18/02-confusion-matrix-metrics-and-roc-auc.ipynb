### Mathematical Background

For imbalanced binary classification (Section 18.4), accuracy alone is misleading. Instead we compute:

$$\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}}, \qquad \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}}, \qquad F_1 = \frac{2 \cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$

In Section 18.4.1 (Worked Example): $\text{TP}=40, \text{FP}=10, \text{FN}=10$ gives $\text{Precision} = \text{Recall} = F_1 = 0.80$.
