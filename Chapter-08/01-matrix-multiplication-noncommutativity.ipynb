```python
# Self-check tests matching Chapter 8 worked example
A = np.array([[1.0, 2.0], [3.0, 4.0]])
B = np.array([[2.0, 0.0], [1.0, -1.0]])

AB = matmul_from_scratch(A, B)
BA = matmul_from_scratch(B, A)

assert np.allclose(AB, [[4.0, -2.0], [10.0, -4.0]])
assert np.allclose(BA, [[2.0, 4.0], [-2.0, -2.0]])
assert np.allclose(AB, A @ B)
assert not np.allclose(AB, BA), "A @ B and B @ A should differ!"

# 3x3 non-commutativity check
rng = np.random.default_rng(42)
M1 = rng.integers(-3, 4, size=(3, 3)).astype(float)
M2 = rng.integers(-3, 4, size=(3, 3)).astype(float)
assert np.allclose(matmul_from_scratch(M1, M2), M1 @ M2)
assert not np.allclose(M1 @ M2, M2 @ M1)

# Incompatible dimensions raise ValueError
try:
    matmul_from_scratch(np.ones((2, 3)), np.ones((2, 2)))
    assert False, "Expected ValueError for incompatible dimensions"
except ValueError:
    pass

# Batched neural-network layer
X = np.array([[1.0, 2.0, 3.0], [0.0, -1.0, 1.0]])  # (2, 3)
W = np.array([[1.0, 0.0, -1.0], [2.0, 1.0, 0.0]])  # (2, 3) -> 2 outputs
b = np.array([0.5, -0.5])
Y = linear_layer_batch(X, W, b)
assert Y.shape == (2, 2)
assert np.allclose(Y, [[-1.5, 3.5], [-0.5, -1.5]])

print("All Chapter 8.1 checks passed!")
```
