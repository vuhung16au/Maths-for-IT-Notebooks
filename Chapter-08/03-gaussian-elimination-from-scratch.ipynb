## Mathematical Context

Gaussian elimination transforms $[A \mid \mathbf{b}]$ into upper-triangular (row-echelon) form using elementary row operations:
1. **Partial pivoting:** For column $k$, swap row $k$ with the row $r \ge k$ having the largest $|U_{r,k}|$ to avoid division by zero and reduce rounding error.
2. **Elimination:** For each row $i > k$, subtract factor $U_{i,k} / U_{k,k}$ times row $k$ from row $i$.
3. **Back-substitution:** Solve from the bottom row $n-1$ up to row $0$:
