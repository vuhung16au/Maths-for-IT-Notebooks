### Mathematical Background

For $f : \mathbb{R}^n \to \mathbb{R}$, the **gradient vector** $\nabla f(\mathbf{x}) = (\partial f / \partial x_1, \dots, \partial f / \partial x_n)^T$ points in the direction of steepest ascent. Gradient descent steps in the opposite direction:

$$\mathbf{x}_{k+1} = \mathbf{x}_k - \alpha\nabla f(\mathbf{x}_k)$$

In Section 16.1.1 (Worked Example): for $f(x, y) = x^2 + y^2$, $\nabla f(x, y) = (2x, 2y)$. From $\mathbf{x}_0 = (2, -1)$ with $\alpha = 0.1$, one step gives $\mathbf{x}_1 = (1.6, -0.8)$.
