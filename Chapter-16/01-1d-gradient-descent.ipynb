### Mathematical Background

When solving $f'(x) = 0$ algebraically is impractical, **1D gradient descent** minimises $f(x)$ iteratively starting from an initial guess $x_0$:

$$x_{k+1} = x_k - \alpha\,f'(x_k)$$

where $\alpha > 0$ is the **learning rate** (step size).

In Section 16.1.1 (Worked Example): for $f(x) = (x - 3)^2 + 2$, we have $f'(x) = 2(x - 3)$. Starting at $x_0 = 0$ with $\alpha = 0.25$:
- $x_1 = 0 - 0.25(-6) = 1.5$
- $x_2 = 1.5 - 0.25(-3) = 2.25$
Each step halves the distance to the minimum $x^* = 3$.
