### Mathematical Background

A point $\mathbf{x}^*$ where $\nabla f(\mathbf{x}^*) = \mathbf{0}$ is a **stationary point** (Section 16.2). The second-order curvature is captured by the **Hessian matrix** $H_{ij} = \frac{\partial^2 f}{\partial x_i \partial x_j}$:

- All eigenvalues $\lambda_i > 0$: **local minimum**
- All eigenvalues $\lambda_i < 0$: **local maximum**
- Mixed-sign eigenvalues (some $> 0$, some $< 0$): **saddle point**
