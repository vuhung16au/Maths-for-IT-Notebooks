### Mathematical Context

Given a public prime modulus $p$ and generator $g$:
1. Alice chooses secret $a$ and sends $A = g^a \bmod p$.
2. Bob chooses secret $b$ and sends $B = g^b \bmod p$.
3. Both compute the shared secret:
$$s = B^a \bmod p = (g^b)^a \bmod p = g^{ab} \bmod p = (g^a)^b \bmod p = A^b \bmod p.$$
