### Mathematical Context

$$\mathrm{FWER} = 1 - (1 - \alpha)^m.$$

For $\alpha = 0.05$ and $m = 10$, $\mathrm{FWER} \approx 0.401$; for $m = 20$, $\mathrm{FWER} \approx 0.642$!
