### Mathematical Context

For a one-proportion test of $H_0: p = p_0$ given $x$ successes in $n$ trials ($\widehat p = x/n$):
$$\operatorname{SE}_0 = \sqrt{\frac{p_0(1 - p_0)}{n}}, \qquad z = \frac{\widehat p - p_0}{\operatorname{SE}_0}.$$

In the chapter's worked example ($60/100$ clicks vs. $p_0 = 0.5$), $\operatorname{SE}_0 = 0.05$, $z = 2.0$, and the two-sided $p$-value is $\operatorname{erfc}(|z|/\sqrt{2}) \approx 0.0455$.

$$\operatorname{SE}_{\text{pooled}} = \sqrt{\widehat p(1 - \widehat p)\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}, \qquad z = \frac{\widehat p_1 - \widehat p_2}{\operatorname{SE}_{\text{pooled}}}.$$
