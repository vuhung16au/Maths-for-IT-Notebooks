### Mathematical Context

- **Type I error ($\alpha$):** rejecting $H_0$ when $H_0$ is true ($p_1 = p_2$).
- **Type II error ($\beta$):** failing to reject $H_0$ when $H_1$ is true ($p_1 \neq p_2$).
- **Statistical power ($1 - \beta$):** probability of rejecting $H_0$ when a real effect exists.

To measure practical magnitude independently of sample size $n$, **Cohen's $h$** for two proportions is:
$$h = 2\arcsin(\sqrt{p_1}) - 2\arcsin(\sqrt{p_2}).$$
Rules of thumb: $|h| \approx 0.2$ (small), $0.5$ (medium), $0.8$ (large).
