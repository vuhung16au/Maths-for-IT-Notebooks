### Mathematical Context: Quantifiers, Implication & Counterexamples

#### Predicates and Quantifiers (Section 1.2)
A **predicate** $P(x)$ is a boolean-valued function defined over a domain $S$.
- **Universal Quantifier ($\forall x \in S, P(x)$):** True if and only if $P(x)$ holds for *every* $x \in S$.
- **Existential Quantifier ($\exists x \in S, P(x)$):** True if $P(x)$ holds for *at least one* $x \in S$.
- **Quantifier Negation Duality:**
  $$\neg(\forall x \in S, P(x)) \iff \exists x \in S, \neg P(x)$$
  To disprove a universal claim, finding **one single counterexample** is necessary and sufficient.
  *Example (Exercise 1.2):* The negation of "every server in the cluster is available" is "at least one server in the cluster is unavailable."
  *Example (Exercise 1.5):* Disprove "every odd integer is prime". The integer $9$ is odd ($2 \cdot 4 + 1$) but composite ($3 \times 3$), disproving the universal claim.

#### Material Implication and Forms of Conditional Statements (Section 1.4)
The implication $P \implies Q$ asserts: "if $P$ is true, then $Q$ must be true."
- **Equivalence:** $P \implies Q \equiv \neg P \lor Q$.
- **Vacuous Truth:** When $P$ is False, $P \implies Q$ is automatically **True**, regardless of $Q$.
- **Contrapositive:** $\neg Q \implies \neg P$. Always logically equivalent to $P \implies Q$.
- **Converse:** $Q \implies P$. **NOT** logically equivalent to $P \implies Q$.
  *Example (Section 1.4, Figure 1.6):* "If $n$ is divisible by 4, then $n$ is even" is True. But its converse "if $n$ is even, then $n$ is divisible by 4" is False ($n = 2$ is even, but $2 \notin 4\mathbb{Z}$).
