### Mathematical Context: De Morgan's Laws & Logical Equivalence

Two boolean expressions $f$ and $g$ over variables $x_1, \dots, x_n$ are **logically equivalent** (denoted $f \equiv g$) if and only if they evaluate to identical truth values for every possible valuation:
$$f \equiv g \iff \forall \mathbf{v} \in \{T, F\}^n, \quad f(\mathbf{v}) = g(\mathbf{v})$$

#### De Morgan's Laws
Augustus De Morgan formulated two fundamental duality laws relating conjunction, disjunction, and negation:
1. **Negation of Conjunction:**
   $$\neg(P \land Q) \iff (\neg P \lor \neg Q)$$
   "Not both $P$ and $Q$" is true if and only if at least one is false.
2. **Negation of Disjunction:**
   $$\neg(P \lor Q) \iff (\neg P \land \neg Q)$$
   "Neither $P$ nor $Q$" is true if and only if both are false.
