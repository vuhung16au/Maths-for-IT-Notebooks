### Mathematical Context: Propositions, Connectives & Truth Tables

In mathematical logic, a **proposition** is a declarative statement that is deterministically either **True** ($T$) or **False** ($F$). Compound propositions are constructed using fundamental logical connectives:
- **Conjunction ($\land$ / AND):** $P \land Q$ is true if and only if both $P$ and $Q$ are true.
- **Disjunction ($\lor$ / OR):** $P \lor Q$ is true if at least one operand is true (inclusive OR).
- **Negation ($\neg$ / NOT):** $\neg P$ reverses the truth valuation of $P$.

#### Section 1.1 Truth Table:
| $P$ | $Q$ | $\neg P$ | $P \land Q$ | $P \lor Q$ | $P \Rightarrow Q$ |
| :---: | :---: | :---: | :---: | :---: | :---: |
| T | T | F | T | T | T |
| T | F | F | F | T | F |
| F | T | T | F | T | T |
| F | F | T | F | F | T |
