### Mathematical Context: Set Operations, Subsets & Filtering

A **set** is an unordered collection of distinct elements. The relationship between elements and sets is governed by precise operations:
- **Membership ($x \in A$):** Indicates that $x$ is an element of $A$.
- **Subset ($A \subseteq B$):** Every element of $A$ is also in $B$: $\forall x \in A, x \in B$.
- **Intersection ($A \cap B$):** $\{x \mid x \in A \land x \in B\} \longleftrightarrow \texttt{a \& b}$
- **Union ($A \cup B$):** $\{x \mid x \in A \lor x \in B\} \longleftrightarrow \texttt{a | b}$
- **Set Difference ($A \setminus B$):** $\{x \mid x \in A \land x \notin B\} \longleftrightarrow \texttt{a - b}$
- **Symmetric Difference ($A \Delta B$):** $(A \setminus B) \cup (B \setminus A) \longleftrightarrow \texttt{a \^{} b}$

<summary>💡 Hint: Non-destructive set operators</summary>
