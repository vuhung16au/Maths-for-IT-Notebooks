### Mathematical Context

For an outcome of probability $p > 0$, its **information content** (surprise) in bits is:
$$I(p) = -\log_2 p = \log_2\frac{1}{p}.$$

For a discrete distribution $P = (p_1, \dots, p_k)$, **Shannon entropy** is the expected information:
$$H(P) = -\sum_{i=1}^k p_i \log_2 p_i,$$
with the convention $0 \log_2 0 = 0$.

- Fair coin $P = (0.5, 0.5) \implies H(P) = 1.0\text{ bit}$.
